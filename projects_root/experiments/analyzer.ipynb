```python
with open(path, 'rb') as f:
    stats = pickle.load(f)

task_statman_name, task_stats = 'task_stats', stats['task_stats']
a_statman_names, agents_stats = [statman_name for statman_name in stats.keys() if 'agent' in statman_name], [stats[statman_name] for statman_name in stats.keys() if 'agent' in statman_name]
all_stats = [task_stats] + agents_stats


def print_statman_names(stats):
    print('-'*100)
    print(f'N statmanagers: {len(stats)}')
    for statman_name in stats:
        print(f'statman name: {statman_name}')
    print('-'*100)

def print_stat_names(stats):
    
    for statman_name in stats:
        print('-'*100)
        print(f'statman name: {statman_name}')
        stat_names = stats[statman_name].keys()
        for stat_name in stat_names:
            print(f'    stat name: {stat_name}')
        print('-'*100)

def print_stat_counts(stats):
    for statman_name in stats:
        print('-'*100)
        print(f'statman name: {statman_name}')
        stat_names = stats[statman_name].keys()
        for stat_name in stat_names:
            print(f'    stat name: {stat_name}')
            print(f'    N records: {len(stats[statman_name][stat_name])}')
        print('-'*100)

def print_stat_counts_with_examples(stats):
    for statman_name in stats:
        print('-'*100)
        space = '    '
        print(f'statman name: {statman_name}')
        stat_names = stats[statman_name].keys()
        ind2 = f'{space}{space}'
        ind3 = f'{ind2}{space}'
        for stat_name in stat_names:
            print(f'{ind2}stat name: {stat_name}')
            n_records = len(stats[statman_name][stat_name])
            print(f'{ind2}N records: {n_records}')
            example_indices = [0,n_records//2,n_records-1]
            examples_headers = ['first','middle','last']
            # print(f'{space}{space}examples:{example_indices}')
            for i in range(len(example_indices)):
                example_idx = example_indices[i]
                stat_keys, stat_vals = stats[statman_name][stat_name][example_idx]                
                print(f'{ind2}{examples_headers[i]}({example_idx}):')
                print(f'{ind3}K:{stat_keys}')
                print(f'{ind3}V:{stat_vals}')
        
        print('-'*100)

def info(stats,verbosity=1):

    if verbosity == 1:
        print_statman_names(stats)
    if verbosity == 2:
        print_statman_names(stats)
        print_stat_names(stats)
    if verbosity == 3:
        print_stat_counts(stats)
    if verbosity == 4:
        print_stat_counts_with_examples(stats)


info(stats,verbosity=4)



```

```python
def analyze_bin_task_summary(task_stats,sim_config,verbosity=0):


    picks_kv = get_stat_from_statman(task_stats, 'n_picks')
    drops_kv = get_stat_from_statman(task_stats, 'n_drops')
    picks_k, picks_v = split_task_to_k_v(picks_kv)
    drops_k, drops_v = split_task_to_k_v(drops_kv)

    final_picked_vals = picks_v[-1]
    final_picked_keys = picks_k[-1]
    final_placed_vals = drops_v[-1]
    final_placed_keys = drops_k[-1]

    if verbosity > 0:
        print('-'*100)
        print(f'---{sim_config["task_type"]} task summary ---\n')
        print(f'Simulation timeouts:')
        for k in ['tsys','tphysics','w_step']:
            if k in final_picked_keys:
                print(f'{k}: {final_picked_keys[k]}')
            else:
                print(f'{k}: NA')
        print('-'*100)
        print(f'final_picked_vals: {final_picked_vals}')
        print(f'final_picked_keys: {final_picked_keys}')
        print(f'final_placed_vals: {final_placed_vals}')
        print(f'final_placed_keys: {final_placed_keys}')

    if sim_config['task_type'] == 'bin':
        arm_idx = 0
        total_picks = 0
        total_placed = 0
        for a_idx in range(len(final_picked_vals)):
            for link_name in final_picked_vals[a_idx]:
                
                total_picks_for_arm = final_picked_vals[a_idx][link_name] # valid result for both centralized and decentralized
                total_placed_for_arm = final_placed_vals[a_idx][link_name] # valid result for both centralized and decentralized
                if verbosity > 0:
                    print(f'Arm index: {arm_idx}')
                    print(f'Total picks: {total_picks_for_arm}')
                    print(f'Total placed: {total_placed_for_arm}')
                arm_idx += 1
                
            print('-'*100)
            total_picks += total_picks_for_arm
            total_placed += total_placed_for_arm
        
        if verbosity > 0:
            print(f'Algorithm Total picks: {total_picks}')
            print(f'Algorithm Total placed: {total_placed}')
            print('-'*100)

    ans = {'n_picked':total_picks,'n_placed':total_placed}
    return ans
```
