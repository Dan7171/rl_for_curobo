```python
import matplotlib.pyplot as plt
import numpy as np
import paper_style  # apply the style automatically

x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="Planner A")
plt.plot(x, y2, label="Planner B")
plt.xlabel("Time [s]")
plt.ylabel("Value")
plt.legend()
plt.title("Comparison of planners")

plt.savefig("comparison.pdf")  # sharp, vector graphic
plt.show()
```
